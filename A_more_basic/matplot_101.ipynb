# Get_Started

```python
# 引入
import matplotlib.pyplot as plt
```

```python
# Step1: 基础画图

# x轴
foo_dataset = [1,2,3,4,5]

# y轴
y_1 = [2,4,6,8,10]
y_2 = [1,2,3,4,5]

plt.plot(foo_dataset,y_1)
plt.plot(foo_dataset,y_2)

plt.show()
```

```python
# Step2: 格式，标题，图例，标注

foo_dataset = [1,2,3,4,5]

y_1 = [2,4,6,8,10]
y_2 = [1,2,3,4,5]

plt.plot(foo_dataset,y_1,"ro--",label="abc_label")
plt.plot(foo_dataset,y_2,"b^--",label="def_label")

# titile
plt.title("I am a TITLE")

# label
plt.xlabel("xxx")
plt.ylabel("yyy")

#图例
plt.legend()
plt.show()
```

```python
# Step3: 辅助线

foo_dataset = [1,2,3,4,5]

y_1 = [2,4,6,8,10]
y_2 = [1,2,3,4,5]

plt.plot(foo_dataset,y_1,"ro--",label="abc_label")
plt.plot(foo_dataset,y_2,"b^--",label="def_label")

# titile
plt.title("I am a TITLE")

# label
plt.xlabel("xxx")
plt.ylabel("yyy")

#图例
plt.legend()

# 手动划分网格
plt.xticks([1,2,3,4,5])
plt.yticks([4,8,12])

# 显示网格
plt.grid()


plt.show()
```

```python
# Step4: 限制局部

foo_dataset = [1,2,3,4,5]

y_1 = [2,4,6,8,10]
y_2 = [1,2,3,4,5]

plt.plot(foo_dataset,y_1,"ro--",label="abc_label")
plt.plot(foo_dataset,y_2,"b^--",label="def_label")

# titile
plt.title("I am a TITLE")

# label
plt.xlabel("xxx")
plt.ylabel("yyy")

#图例
plt.legend()

# 手动划分网格
plt.xticks([1,2,3,4,5])
plt.yticks([4,8,12])

# 显示网格
plt.grid()

# 划分局部
plt.xlim(4,5)
plt.ylim(4,10)


plt.show()
```

```python
# Step1: 绘制多个图(matlab语法)

# inch 英寸
plt.figure(figsize=(5,10))

# 2行1列的第一个图
plt.subplot(211)
plt.bar([1,2,3,4,5],[2,4,6,8,10])

plt.xlim(0,6)
plt.ylim(0,11)

# 2行1列的第二个图
plt.subplot(212)
plt.plot([1,2,3,4,5],[2,4,6,8,10])

plt.xlim(0,6)
plt.ylim(0,11)

plt.show()
```

```python
# Step1: 绘制多个图(oop语法)

fig,axes = plt.subplots(1,2,figsize=(12,5))

axes[0].plot([1,2,3,4,5],[2,4,6,8,10])
axes[1].bar([1,2,3,4,5],[2,4,6,8,10])


plt.show()
```

```python
# Step2: 更多的绘制(oop语法)

fig,axes = plt.subplots(2,2,figsize=(6,6))

axes[0,0].plot([1,2,3,4,5],[2,4,6,8,10])
axes[0,1].bar([1,2,3,4,5],[2,4,6,8,10])
axes[1,0].scatter([1,2,3,4,5],[4,6,8,2,10],s=[10,50,20,100,30],c=['r','b','c','y','g'])


plt.show()
```

```python
# Step3: 整个图片的控制

fig,axes = plt.subplots(2,2,figsize=(6,6),facecolor="grey",sharex=True,sharey=True)

axes[0,0].plot([1,2,3,4,5],[2,4,6,8,10])
axes[0,1].bar([1,2,3,4,5],[2,4,6,8,10])
axes[1,0].scatter([1,2,3,4,5],[4,6,8,2,10],s=[10,50,20,100,30],c=['r','b','c','y','g'])
# 删除子图
axes[1,1].remove()

# 子图片设置
axes[0,0].set_title("i am small title")
axes[0,1].set_xlabel("xxx")
axes[1,0].set_ylabel("yyy")


# 大图片设置
fig.suptitle("I AM A BIG TITLE")
fig.supxlabel("BIG XXX")
fig.supylabel("BIG YYY")

plt.show()
```

```python
# Step4: 3D图

fig,axes = plt.subplots(2,2,figsize=(6,6),facecolor="grey",sharex=True,sharey=True)

axes[0,0].plot([1,2,3,4,5],[2,4,6,8,10])
axes[0,1].bar([1,2,3,4,5],[2,4,6,8,10])
axes[1,0].scatter([1,2,3,4,5],[4,6,8,2,10],s=[10,50,20,100,30],c=['r','b','c','y','g'])

# 子图片设置
axes[0,0].set_title("i am small title")
axes[0,1].set_xlabel("xxx")
axes[1,0].set_ylabel("yyy")
axes[1,1].remove()

# 大图片设置
fig.suptitle("I AM A BIG TITLE")
fig.supxlabel("BIG XXX")
fig.supylabel("BIG YYY")


# 3D图

import numpy as np
## 生成数据
# Make data
X = np.array([-2,0,2])
Y = np.array([-2,0,2])

X,Y=np.meshgrid(X,Y)
Z1 = np.ones_like(X)
Z2 = np.zeros_like(Y)

# 位置224
ax = fig.add_subplot(2,2,4,projection='3d',facecolor="grey")
ax.plot_surface(X, Y, Z1)
ax.plot_surface(X, Y, Z2)



plt.show()
```

```python
# Step5: 更加精细的控制

fig,axes = plt.subplots(2,2,figsize=(6,6),facecolor="grey",sharex=True,sharey=True)

axes[0,0].plot([1,2,3,4,5],[2,4,6,8,10])
axes[0,1].bar([1,2,3,4,5],[2,4,6,8,10])
axes[1,0].scatter([1,2,3,4,5],[4,6,8,2,10],s=[10,50,20,100,30],c=['r','b','c','y','g'])

# 子图片设置
axes[0,0].set_title("i am small title")
axes[0,1].set_xlabel("xxx")
axes[1,0].set_ylabel("yyy")
axes[1,1].remove()

# 大图片设置
fig.suptitle("I AM A BIG TITLE")
fig.supxlabel("BIG XXX")
fig.supylabel("BIG YYY")


# 3D图

import numpy as np
## 生成数据
# Make data
X = np.array([-2,0,2])
Y = np.array([-2,0,2])

X,Y=np.meshgrid(X,Y)
Z1 = np.ones_like(X)
Z2 = np.zeros_like(Y)

# 位置224
ax = fig.add_subplot(2,2,4,projection='3d',facecolor="grey")
ax.plot_surface(X, Y, Z1)
ax.plot_surface(X, Y, Z2)


# 精细控制(背景颜色和透明度)
axes[0,0].set_facecolor("red")
axes[0,0].patch.set_alpha(0.2)

axes[1,0].set_facecolor("blue")
axes[1,0].patch.set_alpha(0.2)

# 紧凑控制
plt.tight_layout()


plt.show()
```
