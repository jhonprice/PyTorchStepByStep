```python
# 生成图片
from image_classification import generate_dataset
import numpy as np
from sklearn.model_selection import train_test_split;

imgs_all, labels_all = generate_dataset(img_size=10, n_images=1000, binary=False, seed=17)

x_train,x_val,y_train,y_val= train_test_split(imgs_all,labels_all,test_size=0.2,train_size=0.8,random_state=13)

print(x_train.shape,x_val.shape,y_train.shape,y_val.shape)
```

```python
import os
from PIL import Image
import shutil

# 生成目录
# 检查目录是否存在
if os.path.exists("./data"):
    shutil.rmtree('./data')

os.makedirs('./line_data',exist_ok=True)
os.makedirs('./line_data/train/1',exist_ok=True)
os.makedirs('./line_data/train/0',exist_ok=True)
os.makedirs('./line_data/test/1',exist_ok=True)
os.makedirs('./line_data/test/0',exist_ok=True)
os.makedirs('./line_data/all/1',exist_ok=True)
os.makedirs('./line_data/all/0',exist_ok=True)
```

```python
# all
for i in range(300):
    tmp = np.transpose(imgs_all[i], (1, 2, 0))
    tmp = np.squeeze(tmp, axis=2)
    tmp = tmp.astype(np.uint8)
    tmp_img = Image.fromarray(tmp, mode='L')
    # 根据标签选择保存目录
    dirname = './line_data/all/' + str(labels_all[i])
    pathname = dirname + '/' + str(i) + '.png'
    print(pathname)
    tmp_img.save(pathname)
# train
for i in range(240):
    tmp = np.transpose(x_train[i], (1, 2, 0))
    tmp = np.squeeze(tmp, axis=2)
    tmp = tmp.astype(np.uint8)
    tmp_img = Image.fromarray(tmp, mode='L')
    # 根据标签选择保存目录
    dirname = './line_data/train/' + str(y_train[i])
    pathname = dirname + '/' + str(i) + '.png'
    print(pathname)
    tmp_img.save(pathname)
# test
for i in range(60):
    tmp = np.transpose(x_val[i], (1, 2, 0))
    tmp = np.squeeze(tmp, axis=2)
    tmp = tmp.astype(np.uint8)
    tmp_img = Image.fromarray(tmp, mode='L')
    # 根据标签选择保存目录
    dirname = './line_data/test/' + str(y_val[i])
    pathname = dirname + '/' + str(i) + '.png'
    print(pathname)
    tmp_img.save(pathname)
```

```python
# 使用urlretrieve从GitHub下载data目录到本地
import os
import json
from urllib.request import urlopen, urlretrieve

def download_github_dir(api_url, local_base):
    """
    递归下载GitHub目录到本地
    api_url: GitHub API contents URL
    local_base: 本地保存路径
    """
    os.makedirs(local_base, exist_ok=True)
    
    with urlopen(api_url) as resp:
        items = json.loads(resp.read().decode())
    
    for item in items:
        local_path = os.path.join(local_base, item['name'])
        
        if item['type'] == 'dir':
            # 递归处理子目录
            print(f"📁 进入目录: {item['name']}")
            download_github_dir(item['url'], local_path)
        elif item['type'] == 'file':
            # urlretrieve下载文件
            print(f"⬇ 下载: {item['name']} -> {local_path}")
            urlretrieve(item['download_url'], local_path)
            print(f"   ✅ 完成")

# GitHub API地址
api_url = 'https://api.github.com/repos/jhonprice/PyTorchStepByStep/contents/data_generation/data'
local_dir = './data'

print("开始从GitHub下载data目录...")
download_github_dir(api_url, local_dir)
print("🎉 全部下载完成!")
```
